# Python Notes: `map()`, `filter()`, and `reduce()`

These functions process iterables such as lists, tuples, and sets.

## 1. `map()`

`map()` applies a function to every item in an iterable.

### Syntax

```python
map(function, iterable)
```

### Example

```python
numbers = [1, 2, 3, 4]

squares = map(lambda x: x ** 2, numbers)

print(list(squares))
# [1, 4, 9, 16]
```

Using a normal function:

```python
def double(x):
    return x * 2

result = map(double, [1, 2, 3])
print(list(result))
# [2, 4, 6]
```

### With multiple iterables

```python
a = [1, 2, 3]
b = [4, 5, 6]

result = map(lambda x, y: x + y, a, b)

print(list(result))
# [5, 7, 9]
```

`map()` stops when the shortest iterable is exhausted.

---

## 2. `filter()`

`filter()` selects items for which a function returns `True`.

### Syntax

```python
filter(function, iterable)
```

### Example

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

print(list(even_numbers))
# [2, 4, 6]
```

Using a normal function:

```python
def is_positive(x):
    return x > 0

values = [-2, -1, 0, 1, 2]

result = filter(is_positive, values)

print(list(result))
# [1, 2]
```

### Filtering strings

```python
words = ["apple", "", "banana", "", "cherry"]

result = filter(None, words)

print(list(result))
# ['apple', 'banana', 'cherry']
```

When the function is `None`, `filter()` removes falsy values such as:

```python
False, 0, "", None, [], {}
```

---

## 3. `reduce()`

`reduce()` repeatedly combines items into one final value.

It is available in the `functools` module.

### Syntax

```python
from functools import reduce

reduce(function, iterable)
```

### Example: sum

```python
from functools import reduce

numbers = [1, 2, 3, 4]

total = reduce(lambda x, y: x + y, numbers)

print(total)
# 10
```

The calculation happens like this:

```text
(((1 + 2) + 3) + 4) = 10
```

### Example: product

```python
from functools import reduce

numbers = [1, 2, 3, 4]

product = reduce(lambda x, y: x * y, numbers)

print(product)
# 24
```

### Using an initial value

```python
from functools import reduce

numbers = [1, 2, 3]

result = reduce(lambda x, y: x + y, numbers, 10)

print(result)
# 16
```

Calculation:

```text
10 + 1 + 2 + 3 = 16
```

## Comparison

| Function | Purpose | Result |
|---|---|---|
| `map()` | Transform every item | An iterable of transformed values |
| `filter()` | Select matching items | An iterable of selected values |
| `reduce()` | Combine all items | One final value |

## Combined example

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

result = reduce(
    lambda x, y: x + y,
    filter(
        lambda x: x % 2 == 0,
        map(lambda x: x * 2, numbers)
    )
)

print(result)
# 42
```

Process:

```text
map:    [2, 4, 6, 8, 10, 12]
filter: [2, 4, 6, 8, 10, 12]
reduce: 42
```

A clearer list-comprehension version would be:

```python
result = sum(x * 2 for x in numbers if x * 2 % 2 == 0)
```

## Important points

- In Python 3, `map()` and `filter()` return lazy iterator objects.
- Convert them to a list when you need to display or reuse the results:

```python
list(map(...))
list(filter(...))
```

- Iterators are consumed after one use:

```python
result = map(lambda x: x * 2, [1, 2, 3])

print(list(result))  # [2, 4, 6]
print(list(result))  # []
```

- For simple transformations and filtering, list comprehensions are often more readable:

```python
squares = [x ** 2 for x in numbers]
evens = [x for x in numbers if x % 2 == 0]
```

- For sums, products, minimums, and maximums, built-in functions are often preferable to `reduce()`:

```python
sum(numbers)
min(numbers)
max(numbers)
```

### Quick memory trick

```text
map    → modify
filter → choose
reduce → combine
```
