```python
# A2 (Lesson 11: operator overloading)
class V:
    def __init__(self, x): self.x = x
    def __add__(self, other): return V(self.x * other.x)
    def __eq__(self, other): return self.x == other.x
    def __repr__(self): return f'V({self.x})'

print(V(2) + V(5))
print(V(3) == V(3), V(3) != V(4))

# Answer: 
```

```python
# A3 (Lesson 12: properties)
class Temp:
    def __init__(self, c): self._c = c
    @property
    def f(self): return self._c * 9 / 5 + 32
    @property
    def c(self): return self._c
    @c.setter
    def c(self, v):
        if v < -273: raise ValueError('too cold')
        self._c = v

t = Temp(100)
t.c = 0
print(t.f)
t.f = 50
print('done')

# Answer: 
```

```python
# A4 (Lesson 13: exceptions)
def f(x):
    try:
        print('try')
        return 10 // x
    except ZeroDivisionError:
        print('zero')
        return -1
    else:
        print('else')
    finally:
        print('finally')

print(f(0))
print(f(5))

# Answer: 
```
