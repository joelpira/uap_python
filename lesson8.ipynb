# Currying Functions

Currying is a technique which allows new functions to be created from existing functions by binding one or more parameters to a specific value. 

It is a major source of **reuse** of functions in Python which means that functionality can be **written once**, in one place and then reused in multiple other situations.

```python
## generic multiply function and higher-order function to create multipliers
def multiply(a, b):
    return a * b

def multby(func, num):
    return lambda x: func(x, num)

print('call function multiply\n\n')
print('direct calls:\n')
print(multiply(3, 2))
print(multiply(4, 5))
print(multiply(6, 7))


print('\ncalls using multby:\n')
print(multby(multiply, 2)(3))
print(multby(multiply, 5)(4))
print(multby(multiply, 7)(6))

double = multby(multiply, 2)
print(double(3))
print(double(4))
print(double(5))

triple = multby(multiply, 3)
print(triple(3))
print(triple(4))
print(triple(5))
```

# Closures



```python
def increment(num):
    return num + 1
    
def reset_function():
    global increment
    addition = 50
    increment = lambda num: num + addition
    
print(increment(5))
reset_function()
print(increment(5))
```

What happens step by step:

1. <span style='color:orange'>increment</span> starts as a plain function that adds 1.
2. <span style='color:orange'>print(increment(5))</span> → 6 (5 + 1).
3. <span style='color:orange'>reset_function()</span> runs:
    - Creates a local variable <span style='color:orange'>addition = 50</span>.
    - Reassigns the global <span style='color:orange'>increment</span> to a new <span style='color:orange'>lambda: lambda num: num + addition</span>.
4. <span style='color:orange'>print(increment(5))</span> → 55 (5 + 50).

Where the closure is:

The lambda <span style='color:orange'>lambda num: num + addition</span> is the **closure**. It captures **addition** from <span style='color:orange'>reset_function</span>'s local scope. Even though <span style='color:orange'>reset_function()</span> has already returned (and its local frame is gone), the lambda still holds a reference to that addition variable — it "remembers" the environment it was created in.

In other words, <span style='color:orange'>addition</span> is a free variable in the lambda: it's not a parameter, not global, so it must come from the enclosing scope. That captured binding is the closure.

Why it matters here: the <span style='color:orange'>global</span> increment reassignment is what makes the second <span style='color:orange'>print</span> use the new lambda instead of the original function. The closure mechanism is what lets the lambda keep working correctly after <span style='color:orange'>reset_function</span> is done.
