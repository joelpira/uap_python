# Python Properties: Getter, Setter, and Public Interface

A property lets you access a method like an attribute while controlling how its value is read, changed, or deleted.

It is useful for:

- Validating values
- Hiding internal implementation details
- Keeping a stable public interface
- Calculating values dynamically
- Preventing direct modification of internal data

## 1. Basic Getter and Setter

````python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        """Getter for name."""
        return self._name

    @name.setter
    def name(self, value):
        """Setter for name."""
        if not value:
            raise ValueError("Name cannot be empty")
        self._name = value


person = Person("Alice")

print(person.name)      # Calls the getter
person.name = "Bob"     # Calls the setter

print(person.name)      # Bob
````

The user interacts with <code>name</code> like a normal attribute, while the class controls access internally.

## 2. Why Use _name?
The leading underscore indicates that <code>_name</code> is an internal implementation attribute.

```{python}
self._name = name
``` 

The public interface is:

```python
person.name
``` 
The internal storage is:

``` python
person._name
``` 
By convention, users should access <code>name</code>, not <code>_nameM</code>.


## 3. Getter-Only Property
A property without a setter is read-only.

````python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14159 * self.radius ** 2

circle = Circle(5)

print(circle.area)
````

Invalid call

````python
circle.area = 100
````

It raises an <code>AttributeError</code> because <code>area</code> has no setter.

## 4. Property with Validation

````python
class Product:
    def __init__(self, price):
        self.price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price cannot be negative")
        self._price = value
````

The setter ensures that invalid values cannot enter the object.

````python
product = Product(25)

product.price = 30        # Valid
product.price = -10       # ValueError
````

## 5. More Concise Property Definitions
For simple attributes, you can use the built-in <code>property()</code> function:

````python
class Person:
    def __init__(self, name):
        self._name = name

    def get_name(self):
        return self._name

    def set_name(self, value):
        self._name = value

    name = property(get_name, set_name)
````

However, the decorator form is usually easier to read:

````python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        self._name = value
````

## 6. Concise Properties with <code>dataclasses</code>

If no validation or custom behavior is needed, use a normal attribute or a dataclass:

````python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
````

Usage: 

````python
person = Person("Alice", 30)
print(person.name)
````

## 7. Getter, Setter, and Deleter
A property can define three operations:

````python
class User:
    def __init__(self, username):
        self._username = username

    @property
    def username(self):
        return self._username

    @username.setter
    def username(self, value):
        self._username = value

    @username.deleter
    def username(self):
        del self._username


user = User("alice")

print(user.username)   # Getter
user.username = "bob"  # Setter

del user.username      # Deleter

user.username = "john"  # Setter
print(user.username)   # Getter
````

## 10. Common Naming Pattern

````python
class Example:
    def __init__(self, value):
        self.value = value

    @property
    def value(self):
        return self._value

    @value.setter
    def value(self, value):
        self._value = value
````

The property and its backing attribute normally use these names:

<table>
<tr><td>Purpose</td><td>	Name</td></tr>
<tr><td>Public property</td><td>	value</td></tr>
<tr><td>Internal storage</td><td>	_value</td></tr>
<tr><td>Getter</td><td>	value(self)</td></tr>
<tr><td>Setter</td><td>	value(self, new_value)</td></tr>
</table>

## 11. Key Points
- <code>@property</code> defines a getter.
- <code>@property_name.setter</code> defines a setter.
- A property can be accessed like a normal attribute.
- Use <code>_attribute</code> for internal storage by convention.
- Setters are useful for validation.
- A property without a setter is read-only.
- Use regular attributes when no custom behavior is needed.
- Properties help maintain a stable public interface.
- The decorator syntax is generally clearer than <code>property()</code>.
- Keep properties simple and avoid unnecessary boilerplate.
