### Part 1: Easy OOP Exercises
*Focus: Class definition, basic attributes, `self`, instantiating objects, and simple instance methods.*

#### Exercise 1: The `Beverage` Class


*   **Scenario:** You want to represent different drinks served at a café.
*   **Requirements:**
    1. Define a class named `Beverage`.
    2. The `__init__` method should take two parameters: `name` (a string) and an optional `temperature` (an integer or float).
    3. If `temperature` is not provided when creating the object, it should default to `75` degrees Celsius. Store both values as instance attributes (`self.name` and `self.temperature`).
    4. Write an instance method named `is_hot(self)` that returns `True` if the beverage's temperature is `60` degrees or higher, and `False` otherwise.


#### Exercise 2: The `Person` Class


*   **Scenario:** You need to track basic details and status updates for individuals.
*   **Requirements:**
    1. Define a class named `Person`.
    2. The `__init__` method should initialize the attributes `name` (string) and `age` (integer).
    3. Write a method named `birthday(self)` that prints `"Happy birthday, you were [current_age]"`, increments the `age` attribute by `1` in place, and then prints `"You are now [new_age]"`.
    4. Write a method named `is_teenager(self)` that returns `True` if the person's age is under 20 (and over 12), and `False` otherwise.


#### Exercise 3: The `Book` Class


*   **Scenario:** You are building a catalog for a small bookstore.
*   **Requirements:**
    1. Define a class named `Book`.
    2. The `__init__` method should accept `title` (string), `author` (string), and `price` (float).
    3. Write a method named `get_description(self)` that returns a single formatted string: `"Title: [title] by [author]"`.
    4. Write a method named `apply_discount(self, percentage)` that takes a discount percentage (e.g., `10` for 10% off) and permanently reduces the book's price by that percentage.

    

### Part 2: Difficult OOP Exercises


#### Exercise 4: Bank `Account` with Custom Exceptions


*   **Scenario:** You are building a banking transaction class that must prevent illegal negative amounts from being processed.
*   **Requirements:**
    1. Define a custom exception class named `AmountError` that inherits from Python's base `Exception`.
    2. Define a class named `Account`. It should initialize three attributes: `account_number` (string), `account_holder` (string), and `balance` (float).
    3. Write a `deposit(self, amount)` method and a `withdraw(self, amount)` method.
    4. **The Catch:** If a user passes a negative number (e.g., `-50`) as the `amount` to either method, your program must **raise** an `AmountError` containing a descriptive message.


#### Exercise 5: Inheritance & Overdraft Protection


*   **Scenario:** You want to create a special type of checking account that allows the user to overdraw up to a certain threshold, raising an error if they exceed it.
*   **Requirements:**
    1. Define a custom exception named `BalanceError`.
    2. Create a subclass named `CurrentAccount` that inherits from your `Account` class (from Exercise 4).
    3. The `CurrentAccount` class should override `__init__` to accept an additional parameter: `overdraft_limit` (a positive float, e.g., `100.0` representing a limit of -\$100.0). Use `super()` to initialize the base account attributes.
    4. Override the `withdraw(self, amount)` method.
    5. **Overdraft Protection Logic:** If a withdrawal would cause the account balance to fall below the negative `overdraft_limit` threshold, do **not** decrease the balance. Instead, raise a `BalanceError` exception.


#### Exercise 6: Encapsulated `SmartDevice` with `@property`


*   **Scenario:** You are programming a smart home thermostat. The target temperature must be kept within safe operating boundaries (15.0°C to 30.0°C) to prevent the heating unit from breaking.
*   **Requirements:**
    1. Define a class named `SmartDevice`.
    2. During initialization, set a private target temperature attribute `_target_temp` (a float) to `20.0`. (Remind students that starting an attribute with a single underscore is the Pythonic convention to mark it as private).
    3. Use the `@property` decorator to create a getter method for `target_temp` that returns `self._target_temp`.
    4. Use the `@target_temp.setter` decorator to create a setter method.
    5. **Setter Validation Logic:** Before updating the private variable, verify that the new value is an integer or float, and is between `15.0` and `30.0` (inclusive). If it is outside this range, print a warning `"Invalid Temperature!"` (or raise a `ValueError`) and **do not** update the underlying private variable.
