# Python Error and Exception Handling

Error and exception handling lets a program respond to unexpected situations without crashing immediately.

## 1. Errors vs. Exceptions

**Syntax errors** happen when Python cannot understand the code.

```python
if True
    print("Hello")
```

This produces a `SyntaxError`.

**Exceptions** happen while the program is running.

```python
number = 10 / 0
```

This produces a `ZeroDivisionError`.

Common exceptions include:

| Exception | Example cause |
|---|---|
| `ValueError` | Invalid value, such as `int("abc")` |
| `TypeError` | Incompatible data types |
| `NameError` | Using an undefined variable |
| `IndexError` | Invalid list index |
| `KeyError` | Missing dictionary key |
| `FileNotFoundError` | Opening a missing file |
| `ZeroDivisionError` | Dividing by zero |
| `AttributeError` | Missing object attribute |
| `ImportError` | Importing unavailable functionality |

## 2. Basic `try` and `except`

Put code that might fail inside `try`. Handle the error in `except`.

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except ValueError:
    print("Please enter a valid number.")
except ZeroDivisionError:
    print("Number cannot be zero.")
```

Only the first matching `except` block runs.

## 3. Catching Multiple Exceptions

You can handle multiple exceptions in one block:

```python
try:
    value = int(input("Enter a number: "))
    result = 100 / value
except (ValueError, ZeroDivisionError):
    print("Invalid calculation.")
```

Use separate blocks when the responses should be different.

## 4. Accessing the Exception Object

Use `as` to store the exception:

```python
try:
    number = int("hello")
except ValueError as error:
    print(f"Conversion failed: {error}")
```

The exception object contains information about what went wrong.

## 5. `else` Block

The `else` block runs only if no exception occurs.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid number.")
else:
    print(f"You entered {number}.")
```

This is useful for separating successful code from error-handling code.

## 6. `finally` Block

The `finally` block always runs, whether an exception occurs or not.

```python
try:
    file = open("data.txt")
    content = file.read()
except FileNotFoundError:
    print("File not found.")
finally:
    print("Finished attempting to read the file.")
```

Typical uses include:

- Closing files
- Releasing resources
- Closing database connections
- Cleaning up temporary data

The general structure is:

```python
try:
    ...
except SomeError:
    ...
else:
    ...
finally:
    ...
```

## 7. Handling Files Safely

Use `with` when working with files. It automatically closes the file.

```python
try:
    with open("data.txt", "r") as file:
        content = file.read()
except FileNotFoundError:
    print("The file does not exist.")
```

This is usually better than manually using `finally` to close the file.

## 8. Raising Exceptions

Use `raise` to create an exception deliberately.

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age
```

Usage:

```python
try:
    set_age(-5)
except ValueError as error:
    print(error)
```

`raise` is useful when a function receives invalid input.

## 9. Re-raising an Exception

An exception can be caught, logged, and raised again:

```python
try:
    number = int("abc")
except ValueError:
    print("Logging the error")
    raise
```

The bare `raise` preserves the original exception and traceback.

## 10. Custom Exceptions

Create a custom exception by inheriting from `Exception`.

```python
class InsufficientFundsError(Exception):
    pass
```

Use it in a class:

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError("Not enough money")
        self.balance -= amount
```

Handle it:

```python
account = BankAccount(100)

try:
    account.withdraw(150)
except InsufficientFundsError as error:
    print(error)
```

Custom exceptions make application-specific errors clearer.

## 11. Exception Inheritance

Exceptions are organized in a hierarchy.

```python
BaseException
└── Exception
    ├── ValueError
    ├── TypeError
    ├── OSError
    └── RuntimeError
```

Catching a parent exception also catches its child exceptions:

```python
try:
    value = int("abc")
except Exception:
    print("An exception occurred")
```

However, avoid catching `Exception` unless you have a good reason. Specific exceptions are safer and clearer.

## 12. Avoid Bare `except`

Avoid this:

```python
try:
    risky_operation()
except:
    print("Something went wrong")
```

It can hide serious problems, including interruptions and system-exiting exceptions.

Prefer:

```python
try:
    risky_operation()
except ValueError:
    print("Invalid value")
```

If a broad handler is necessary:

```python
try:
    risky_operation()
except Exception as error:
    print(f"Unexpected error: {error}")
```

## 13. Exception Chaining

Use `from` to show that one exception was caused by another:

```python
def load_number():
    try:
        return int("abc")
    except ValueError as error:
        raise RuntimeError("Could not load number") from error
```

This preserves the original cause and makes debugging easier.

## 14. Assertions

`assert` checks that a condition is true.

```python
age = 20
assert age >= 0
```

If the condition is false, Python raises `AssertionError`:

```python
age = -1
assert age >= 0, "Age cannot be negative"
```

Assertions are mainly for internal debugging and program assumptions. Do not use them as the main way to validate user input.

Use this instead for user input:

```python
if age < 0:
    raise ValueError("Age cannot be negative")
```

## 15. Returning Error Values vs. Raising Exceptions

For expected results, returning a value may be appropriate:

```python
def divide(a, b):
    if b == 0:
        return None
    return a / b
```

For invalid conditions that should be handled separately, raise an exception:

```python
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return a / b
```

Use exceptions when the caller should be required to handle the problem.

## 16. Practical Example

```python
def get_discounted_price(price, discount):
    if not isinstance(price, (int, float)):
        raise TypeError("Price must be a number")

    if not isinstance(discount, (int, float)):
        raise TypeError("Discount must be a number")

    if price < 0:
        raise ValueError("Price cannot be negative")

    if not 0 <= discount <= 100:
        raise ValueError("Discount must be between 0 and 100")

    return price * (1 - discount / 100)


try:
    final_price = get_discounted_price(100, 20)
except (TypeError, ValueError) as error:
    print(f"Error: {error}")
else:
    print(f"Final price: {final_price}")
```

Output:

```text
Final price: 80.0
```

## 17. Best Practices

- Catch specific exceptions.
- Keep the `try` block small.
- Use `else` for code that should run after success.
- Use `finally` for cleanup.
- Use `with` for files and other resources.
- Raise meaningful exceptions.
- Create custom exceptions for domain-specific errors.
- Do not silently ignore exceptions.
- Avoid using exceptions for ordinary control flow.
- Preserve the original cause with exception chaining.
- Use logging instead of `print()` in larger applications.

A concise pattern to remember is:

```python
try:
    operation()
except SpecificError as error:
    handle_error(error)
else:
    handle_success()
finally:
    clean_up()
```
