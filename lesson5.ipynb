# Recursion

A recursive solution in a programming language such as Python is one in which a
function calls itself one or more times in order to solve a particular problem.

``` python
def fact(n):
    if n == 0:          # base case
        return 1
    return n * fact(n-1)  # recursive step

print(fact(5))  # 120

```

### Key rules for correct recursion
- Base case: must be reachable for all valid inputs.
- Progress toward base case: each recursive call should move “closer” to stopping.
- Avoid deep recursion: Python’s recursion limit is relatively low.


``` python
def total(nums, i=0):
    if i == len(nums):   # base case
        return 0
    return nums[i] + total(nums, i+1)
```

### Common problems
- Infinite recursion: base case never reached.
- Wrong parameters: recursive call doesn’t change the state properly.
- Too deep recursion: for large inputs, prefer an iterative loop or a stack-based approach.


````python
## working with recursion
def fact(n):
    if n == 0:          # base case
        return 1
    return n * fact(n-1)  # recursive step

print(fact(5))  # 120
````

````python
def total(nums, i=0):
    if i == len(nums):  # base case
        return 0
    return nums[i] + total(nums, i + 1)

print(total([1, 2, 3, 4, 5]))  # 15
print(total(''))  
````

## Examples of **bad recursion**

### 1) No base case → infinite recursion
```python
def bad(n):
    return bad(n)  # never changes, never stops

print(bad(5))
```
This will quickly raise `RecursionError` because the function never reaches a stopping condition.

### 2) Base case exists, but you never move toward it
```python
def bad(n):
    if n == 0:
        return 1
    return bad(n)  # should have used n-1 or something, but doesn't

print(bad(5))
```
Even though `n == 0` is a base case, the recursive call never changes `n`, so it never reaches `0`.

### 3) State changes the wrong way
```python
def bad(n):
    if n == 0:
        return 0
    return bad(n + 1)  # moves away from 0

print(bad(5))
```
This moves farther from the stopping condition forever.

### 4) Wrong branching: always recursing, even when input is “done”
```python
def bad(n):
    if n == 0:
        return 0
    return bad(n-1) + bad(n-1)  # doubles work each level (explodes fast)

print(bad(30))
```
This isn’t “infinite” recursion, but it’s very inefficient because it makes **two recursive calls** each time, causing a rapid explosion in calls.
