# Python Collections, Tuples, and Lists

A **collection** stores multiple values in one object. Python provides several built-in collection types:

- Lists
- Tuples
- Sets
- Dictionaries

The main differences are mutability, ordering, and duplicate handling.

| Type | Ordered | Changeable | Allows duplicates | Syntax |
|---|---:|---:|---:|---|
| List | Yes | Yes | Yes | `[1, 2, 3]` |
| Tuple | Yes | No | Yes | `(1, 2, 3)` |
| Set | No guaranteed index order | Yes | No | `{1, 2, 3}` |
| Dictionary | Insertion ordered | Yes | Keys must be unique | `{"a": 1}` |

## 1. Lists

A **list** is an ordered and changeable collection.

```python
numbers = [10, 20, 30]
names = ["Alice", "Bob", "Charlie"]
mixed = [10, "Python", True, 3.14]
```

Lists can contain values of different types, although using one main type is often clearer.

## 2. Accessing List Items

List indexes start at `0`.

```python
names = ["Alice", "Bob", "Charlie"]

print(names[0])   # Alice
print(names[1])   # Bob
print(names[-1])  # Charlie
```

Negative indexes count from the end.

## 3. List Slicing

Slicing extracts part of a list.

```python
numbers = [0, 1, 2, 3, 4, 5]

print(numbers[1:4])  # [1, 2, 3]
print(numbers[:3])   # [0, 1, 2]
print(numbers[3:])   # [3, 4, 5]
print(numbers[::2])  # [0, 2, 4]
```

The general syntax is:

```python
list[start:stop:step]
```

The `stop` index is not included.

## 4. Changing List Items

Lists are mutable, so their values can be changed.

```python
numbers = [10, 20, 30]

numbers[1] = 99

print(numbers)  # [10, 99, 30]
```

Change multiple items with slicing:

```python
numbers[0:2] = [1, 2]
```

## 5. Adding Items to Lists

### `append()`

Adds one item to the end.

```python
numbers = [1, 2]
numbers.append(3)

print(numbers)  # [1, 2, 3]
```

### `extend()`

Adds multiple items.

```python
numbers = [1, 2]
numbers.extend([3, 4])

print(numbers)  # [1, 2, 3, 4]
```

### `insert()`

Adds an item at a specific position.

```python
numbers = [1, 3]
numbers.insert(1, 2)

print(numbers)  # [1, 2, 3]
```

## 6. Removing List Items

### `remove()`

Removes the first matching value.

```python
numbers = [1, 2, 2, 3]
numbers.remove(2)

print(numbers)  # [1, 2, 3]
```

### `pop()`

Removes and returns an item.

```python
numbers = [10, 20, 30]

last = numbers.pop()
print(last)     # 30
print(numbers)  # [10, 20]
```

Remove by index:

```python
numbers.pop(0)
```

### `del`

Deletes an item or slice.

```python
numbers = [1, 2, 3, 4]

del numbers[1]
del numbers[1:3]
```

### `clear()`

Removes all items.

```python
numbers.clear()
```

## 7. Useful List Methods

```python
numbers = [3, 1, 2, 1]

print(numbers.count(1))  # 2
print(numbers.index(2))  # 2
```

Sorting:

```python
numbers.sort()
print(numbers)  # [1, 1, 2, 3]
```

Reverse sorting:

```python
numbers.sort(reverse=True)
```

Reverse in place:

```python
numbers.reverse()
```

Copying:

```python
copy = numbers.copy()
```

## 8. `sort()` vs. `sorted()`

`sort()` changes the original list:

```python
numbers = [3, 1, 2]
numbers.sort()
```

`sorted()` creates a new sorted list:

```python
numbers = [3, 1, 2]
result = sorted(numbers)

print(numbers)  # [3, 1, 2]
print(result)   # [1, 2, 3]
```

## 9. List Comprehensions

A list comprehension creates a list concisely.

```python
squares = [number ** 2 for number in range(5)]

print(squares)
# [0, 1, 4, 9, 16]
```

With a condition:

```python
even_numbers = [
    number for number in range(10)
    if number % 2 == 0
]
```

Equivalent traditional code:

```python
even_numbers = []

for number in range(10):
    if number % 2 == 0:
        even_numbers.append(number)
```

## 10. Nested Lists

A list can contain other lists.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0][1])  # 2
```

Loop through a nested list:

```python
for row in matrix:
    for value in row:
        print(value)
```

## 11. Tuples

A **tuple** is an ordered and immutable collection.

```python
coordinates = (10, 20)
colors = ("red", "green", "blue")
```

Tuples support indexing and slicing:

```python
print(colors[0])    # red
print(colors[1:])   # ('green', 'blue')
```

## 12. Single-Item Tuples

A comma is required for a one-item tuple.

```python
value = (10,)  # Tuple
not_tuple = (10)  # Integer
```

The comma creates the tuple:

```python
print(type(value))
print(type(not_tuple))
```

## 13. Tuple Immutability

Tuple items cannot be changed after creation.

```python
coordinates = (10, 20)

# coordinates[0] = 50  # TypeError
```

Tuples do not have methods such as `append()` or `remove()`.

You can create a new tuple instead:

```python
coordinates = (50, 20)
```

## 14. Tuple Unpacking

Tuple unpacking assigns tuple values to variables.

```python
person = ("Alice", 30)

name, age = person

print(name)  # Alice
print(age)   # 30
```

Swapping variables:

```python
a = 10
b = 20

a, b = b, a
```

Using `*` to collect remaining values:

```python
numbers = (1, 2, 3, 4, 5)

first, *middle, last = numbers

print(first)   # 1
print(middle)  # [2, 3, 4]
print(last)    # 5
```

## 15. Returning Multiple Values

Python functions can return multiple values using a tuple.

```python
def get_user():
    return "Alice", 30

name, age = get_user()
```

The function actually returns:

```python
("Alice", 30)
```

## 16. Tuple Methods

Tuples have two main methods:

```python
values = (1, 2, 2, 3)

print(values.count(2))  # 2
print(values.index(3))  # 3
```

## 17. Lists vs. Tuples

| Feature | List | Tuple |
|---|---|---|
| Mutable | Yes | No |
| Syntax | `[1, 2]` | `(1, 2)` |
| Methods | Many | Few |
| Use case | Data that changes | Fixed data |
| Can be dictionary key | No | Yes, if contents are hashable |
| Usually more lightweight | No | Yes |

Use a list when the collection needs to change:

```python
shopping_cart = ["book", "pen"]
shopping_cart.append("notebook")
```

Use a tuple for fixed data:

```python
rgb_color = (255, 0, 0)
```

## 18. Sets

A **set** is an unordered collection of unique values.

```python
numbers = {1, 2, 3, 3}

print(numbers)  # {1, 2, 3}
```

Sets automatically remove duplicates.

Create an empty set with:

```python
items = set()
```

This creates an empty dictionary:

```python
items = {}
```

## 19. Set Operations

```python
first = {1, 2, 3}
second = {3, 4, 5}
```

Union:

```python
print(first | second)
# {1, 2, 3, 4, 5}
```

Intersection:

```python
print(first & second)
# {3}
```

Difference:

```python
print(first - second)
# {1, 2}
```

Symmetric difference:

```python
print(first ^ second)
# {1, 2, 4, 5}
```

Useful methods:

```python
first.add(4)
first.remove(1)
first.discard(10)  # Does not raise an error if missing
```

## 20. Dictionaries

A **dictionary** stores key-value pairs.

```python
person = {
    "name": "Alice",
    "age": 30
}
```

Access values by key:

```python
print(person["name"])
```

Use `.get()` to avoid a `KeyError`:

```python
print(person.get("email"))
```

Add or update values:

```python
person["age"] = 31
person["city"] = "London"
```

Remove values:

```python
person.pop("city")
```

Loop through a dictionary:

```python
for key, value in person.items():
    print(key, value)
```

Useful dictionary methods:

```python
person.keys()
person.values()
person.items()
person.get("name")
person.update({"age": 32})
```

## 21. The `collections` Module

Python also provides a standard library module named `collections`. It contains specialized collection types.

### `Counter`

Counts objects.

```python
from collections import Counter

letters = Counter("banana")

print(letters)
# Counter({'a': 3, 'n': 2, 'b': 1})
```

### `defaultdict`

Provides a default value for missing keys.

```python
from collections import defaultdict

groups = defaultdict(list)

groups["fruit"].append("apple")
groups["fruit"].append("banana")

print(groups)
```

### `deque`

A double-ended queue that efficiently adds and removes items from both ends.

```python
from collections import deque

queue = deque(["A", "B"])

queue.append("C")
queue.appendleft("Start")

print(queue)
```

Remove from either end:

```python
queue.pop()
queue.popleft()
```

### `namedtuple`

Creates tuple-like objects with named fields.

```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])

point = Point(10, 20)

print(point.x)
print(point.y)
```

For new code, `dataclass` is often a more flexible alternative.

## 22. Copying Collections

Assignment does not create a new collection.

```python
original = [1, 2, 3]
copy = original

copy.append(4)

print(original)  # [1, 2, 3, 4]
```

Both variables refer to the same list.

Create a shallow copy:

```python
copy = original.copy()
```

Other options:

```python
copy = list(original)
copy = original[:]
```

For nested collections, use `deepcopy()` when necessary:

```python
from copy import deepcopy

copy = deepcopy(original)
```

## 23. Common Collection Functions

```python
numbers = [4, 2, 8, 1]

len(numbers)       # Number of items
min(numbers)       # Smallest item
max(numbers)       # Largest item
sum(numbers)       # Total
sorted(numbers)    # New sorted list
any(numbers)       # True if any item is truthy
all(numbers)       # True if all items are truthy
```

Use `enumerate()` to access indexes while looping:

```python
names = ["Alice", "Bob"]

for index, name in enumerate(names):
    print(index, name)
```

Use `zip()` to combine collections:

```python
names = ["Alice", "Bob"]
ages = [30, 25]

for name, age in zip(names, ages):
    print(name, age)
```

- Use a **tuple** for ordered data that should not change.
- Use a **set** for unique values and fast membership checks.
- Use a **dictionary** for key-value relationships.
- Use `Counter` to count values.
- Use `deque` for efficient queue operations.
- Use `defaultdict` when missing keys need automatic defaults.
