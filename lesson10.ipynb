# Python Classes

## Class Side Data

In Python classes can also have attributes; these are referred to as class variables or
attributes (as opposed to instance variables or attributes).

### Recall

python

``` python
dog1.name       # instance attribute
Dog.species     # class attribute
dog1.species    # inherited/accessed class attribute

```


<strong>Class variable:</strong> shared by the class and its objects.<br/>
<strong>Instance variable:</strong> specific to one object.<br/>
<strong>Attribute:</strong> the general term for data or behavior accessed with dot notation.<br/>

````python
class Person:
    """ An example class to hold a persons name and age"""
    instance_count = 0
    def __init__(self, name, age):
        Person.instance_count += 1
        self.name = name
        self.age = age

"""
The variable instance_count is not part of an individual object, rather it is
part of the class and all instances of the class can access that shared variable by
prefixing it with the class name.
"""

p1 = Person('Jason', 36)
p2 = Person('Carol', 21)
p3 = Person('James', 19)
p4 = Person('Tom', 31)
print('Person.instance_count:', Person.instance_count)
````

## Class Side Methods

Class methods are written in a similar manner to any other method but are
decorated with @classmethod and take a first parameter which represents the
class rather than an individual instance. This decoration is written before the
method declaration.

- @staticmethod: method does not need self or cls.
- @classmethod: method receives the class as cls.
- @property: lets you call a method like an attribute.


<p>cls is the conventional name for the class itself inside a class method. It is similar to self, but:</p>

- <i>self</i> refers to a particular object instance.
- <i>cls</i> refers to the class.
    
``` python

class Example:
    @staticmethod
    def say_hello():
        print("Hello")

    @classmethod
    def create(cls):
        return cls()

    @property
    def description(self):
        return "An example"

```

````python
class Person:
    """ An example class to hold a persons name and age"""
    instance_count = 0

    @classmethod
    def increment_instance_count(cls):
        cls.instance_count += 1

    def __init__(self, name, age):
        Person.increment_instance_count()
        self.name = name
        self.age = age
````

### Explanation
This code defines a `Person` class that stores a person’s name and age while counting how many `Person` objects have been created.

```python
class Person:
```

Defines a class named `Person`. A class is a blueprint for creating objects.

```python
""" An example class to hold a persons name and age"""
```

This is the class’s **docstring**: a description of what the class is used for.

```python
instance_count = 0
```

This is a **class variable**. It belongs to the `Person` class and is shared by all `Person` objects.

It starts at `0` because no objects have been created yet.

```python
@classmethod
def increment_instance_count(cls):
```

`@classmethod` means this method belongs to the class rather than to one specific object.

The parameter `cls` refers to the `Person` class itself.

```python
cls.instance_count += 1
```

This increases the shared class variable by one.

It is equivalent to:

```python
Person.instance_count += 1
```

Using `cls` is more flexible because it also works properly with subclasses.

```python
def __init__(self, name, age):
```

`__init__` is the initializer method. Python calls it automatically whenever a new `Person` object is created.

- `self` refers to the new object.
- `name` and `age` are values provided when creating the object.

```python
Person.increment_instance_count()
```

Calls the class method and increases the number of created `Person` objects.

This could also be written as:

```python
self.increment_instance_count()
```

or:

```python
type(self).increment_instance_count()
```

Using `Person` specifically means the method is called on the `Person` class.

```python
self.name = name
self.age = age
```

These create **instance variables**:

- `self.name` stores the name for this particular object.
- `self.age` stores the age for this particular object.

Example:

```python
person1 = Person("Alice", 30)
person2 = Person("Bob", 25)

print(person1.name)            # Alice
print(person2.name)            # Bob
print(Person.instance_count)   # 2
```

The sequence is:

1. `person1` is created.
2. `instance_count` becomes `1`.
3. Alice’s name and age are stored.
4. `person2` is created.
5. `instance_count` becomes `2`.
6. Bob’s name and age are stored.

In summary:

| Code | Meaning |
|---|---|
| `instance_count` | Shared class variable |
| `cls` | The class itself |
| `self` | One particular object |
| `self.name` | Instance variable |
| `@classmethod` | Marks a method that receives the class as `cls` |
| `__init__` | Runs when a new object is created |

## Static Methods

Static methods are defined within a class but are not tied to either the class nor
any instance of the class; they do not receive the special first parameter representing
either the class (cls for class methods) or the instances (self for instance methods).


````python
class Person:
    @staticmethod
    def static_function():
        print('Static method')

# Static methods are invoked via the name of the class they are defined in, for
# example:
Person.static_function()
````

A **static method** is a method inside a class that does not need access to the instance (`self`) or the class (`cls`). Define it with `@staticmethod`.

```python
class MathTools:

    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def is_even(number):
        return number % 2 == 0
```

You can call static methods directly through the class:

```python
print(MathTools.add(3, 4))       # 7
print(MathTools.is_even(10))     # True
```

You can also call them through an object:

```python
tools = MathTools()

print(tools.add(2, 5))            # 7
```

However, because the method does not use object or class data, calling it through the class is usually clearer.

Here is an example related to the `Person` class:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @staticmethod
    def is_valid_age(age):
        return age >= 0

    def describe(self):
        return f"{self.name} is {self.age} years old."
```

Usage:

```python
print(Person.is_valid_age(25))    # True
print(Person.is_valid_age(-2))    # False

person = Person("Alice", 30)
print(person.describe())           # Alice is 30 years old
```

`is_valid_age()` is static because it only needs the `age` argument. It does not need `self` or `cls`.

The three common method types are:

```python
class Example:
    def instance_method(self):
        pass

    @classmethod
    def class_method(cls):
        pass

    @staticmethod
    def static_method():
        pass
```

- Instance method: receives `self`; works with object data.
- Class method: receives `cls`; works with class data.
- Static method: receives neither automatically; acts like a regular function organized inside a class.
