```python
count = 10
def outer():
    count = 5
    def inner():
        nonlocal count
        count +=3
        print('Inner: ',count)

    inner()
    print('Outer: ',count)

outer()
print('Global ', count)
```

```python
def apply_twice(func, arg):
    # return func(func(arg))
    return func(func(arg))

double = lambda x: x * 2
square = lambda x: x ** 2

print(apply_twice(double, 3))  
print(apply_twice(square, 3))  

```

Class Mechanics and Instance Methods

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def update_score(self, points):
        self.score += points

    def is_passing(self):
        return self.score >= 75


s1 = Student('Alex', 70)

s1.update_score(8)

print(s1.name, 'Passing:', s1.is_passing())
```

Design a complete Python class named BankAccount using primitive data types:

1. <span style='color:orange'>__init__</span> accepts <span style='color:orange'>account_holder (string)</span> and <span style='color:orange'>optional balance (float, default
0.0)</span>.
2. <span style='color:orange'>deposit(amount)</span> increases balance and returns the new balance.
3. <span style='color:orange'>withdraw(amount)</span> checks if amount > self.balance. If true, prints
'Insufficient Funds' and leaves balance unchanged; otherwise reduces balance and returns new balance.
4. __str__ returns 'Account[Holder: {name}, Balance: ${balance:.2f}]'.
Put your answer below:

```python
class BankAccount:
    def __init__(self, name, balance=0.0):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            print('Insufficient Funds')
            return self.balance
        self.balance -= amount
        return self.balance


    def __str__(self):
        ''' Use descriptive string representation for the account. '''
        return f'Account[Holder: {self.name}, Balance: ${self.balance:.2f}]'


b = BankAccount('John Doe', 100.0)
print(b)
```
