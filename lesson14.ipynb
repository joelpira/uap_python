# Python Modules and Packages

## 1. What Is a Module?

A **module** is a Python file containing code such as:

- Variables
- Functions
- Classes
- Statements

A module has a `.py` extension.

Example file: `math_tools.py`

```python
PI = 3.14159

def add(a, b):
    return a + b

def square(number):
    return number ** 2
```

The file can be imported into another Python program.

## 2. Importing a Module

```python
import math_tools

print(math_tools.add(2, 3))
print(math_tools.square(4))
```

Use the module name followed by a dot to access its contents.

```python
module_name.function_name()
module_name.variable_name
```

## 3. Importing Specific Items

```python
from math_tools import add, square

print(add(2, 3))
print(square(4))
```

You can import one or more specific objects from a module.

Avoid importing too many names because it can make code harder to understand.

## 4. Importing with an Alias

Use `as` to create a shorter or alternative name.

```python
import math_tools as mt

print(mt.add(5, 2))
```

Another example:

```python
import datetime as dt

today = dt.date.today()
```

## 5. Importing Everything

```python
from math_tools import *
```

This imports all public names from the module.

It is generally discouraged because:

- It is unclear where names came from.
- Names can accidentally overwrite each other.
- Code becomes harder to maintain.

Prefer explicit imports:

```python
from math_tools import add, square
```

## 6. Standard Library Modules

Python includes many built-in modules.

```python
import math

print(math.sqrt(25))
print(math.pi)
```

Common standard library modules include:

| Module | Purpose |
|---|---|
| `math` | Mathematical operations |
| `random` | Random values |
| `datetime` | Dates and times |
| `os` | Operating system features |
| `sys` | Python interpreter features |
| `json` | JSON data |
| `re` | Regular expressions |
| `pathlib` | File and directory paths |
| `collections` | Specialized data structures |

Example:

```python
import random

number = random.randint(1, 10)
print(number)
```

## 7. Creating Your Own Module

File: `calculator.py`

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
```

File: `main.py`

```python
import calculator

print(calculator.add(10, 5))
print(calculator.subtract(10, 5))
```

Both files should normally be in the same directory.

## 8. The `__name__` Variable

Every Python module has a special variable called `__name__`.

When a file is run directly:

```python
__name__ == "__main__"
```

When the file is imported:

```python
__name__ == "module_name"
```

Use this pattern to separate reusable code from test code:

```python
def greet(name):
    return f"Hello, {name}!"

if __name__ == "__main__":
    print(greet("Alice"))
```

The test code runs only when the file is executed directly, not when it is imported.

## 9. What Is a Package?

A **package** is a directory containing related Python modules.

Example structure:

```text
project/
│
├── main.py
└── tools/
    ├── __init__.py
    ├── calculator.py
    └── formatting.py
```

The `tools` directory is a package, and the files inside it are modules.

## 10. Importing from a Package

```python
from tools import calculator

print(calculator.add(2, 3))
```

Or import directly from a module:

```python
from tools.calculator import add

print(add(2, 3))
```

The dot represents a package or module level:

```python
package.module.object
```

## 11. The `__init__.py` File

`__init__.py` marks a directory as a Python package and can contain package setup code.

Example:

```python
# tools/__init__.py
```

It can also expose selected functions:

```python
from .calculator import add, subtract
```

Then users can write:

```python
from tools import add, subtract
```

Instead of:

```python
from tools.calculator import add, subtract
```

Modern Python can support packages without `__init__.py`, but including the file is still common and useful.

## 12. Relative Imports

Inside a package, use relative imports to import from nearby modules.

File structure:

```text
tools/
├── __init__.py
├── calculator.py
└── report.py
```

Inside `report.py`:

```python
from .calculator import add
```

The dot means “from the current package.”

Two dots mean the parent package:

```python
from ..common import utilities
```

Relative imports are generally used inside package code.

## 13. Absolute Imports

An absolute import starts from the project or package root.

```python
from tools.calculator import add
```

Absolute imports are often easier to understand because they show the complete path.

## 14. Modules vs. Packages

| Feature | Module | Package |
|---|---|---|
| Meaning | A single Python file | A directory of related modules |
| Example | `calculator.py` | `tools/` |
| Contains | Functions, classes, variables | Modules and subpackages |
| Import example | `import calculator` | `from tools import calculator` |

## 15. Third-Party Packages

Third-party packages are created by external developers and installed separately.

Install a package with:

```bash
python -m pip install requests
```

Use it in Python:

```python
import requests

response = requests.get("https://example.com")
print(response.status_code)
```

A package only needs to be installed once in a particular Python environment.

## 16. Requirements Files

A `requirements.txt` file lists project dependencies.

Example:

```text
requests==2.32.0
numpy>=1.26
```

Install all dependencies with:

```bash
python -m pip install -r requirements.txt
```

## 17. Virtual Environments

A virtual environment keeps a project’s packages separate from other projects.

Create one:

```bash
python -m venv .venv
```

Activate it on Windows:

```bash
.venv\Scripts\activate
```

Activate it on macOS or Linux:

```bash
source .venv/bin/activate
```

Install packages inside the environment:

```bash
python -m pip install requests
```

Deactivate it:

```bash
deactivate
```

## 18. Import Errors

If Python cannot find a module, it raises:

```python
ModuleNotFoundError
```

Example:

```python
import unknown_module
```

Possible causes:

- The module is not installed.
- The file is not in the correct directory.
- The package name is misspelled.
- The virtual environment is not activated.
- The import path is incorrect.

## 19. Useful Module Attributes

Python modules have special attributes.

```python
import math

print(math.__name__)
print(math.__doc__)
print(math.__file__)
```

Common attributes include:

| Attribute | Meaning |
|---|---|
| `__name__` | Module’s name |
| `__doc__` | Module documentation |
| `__file__` | Module’s file location |
| `__package__` | Module’s package |

## 20. Best Practices

- Keep related code in the same module.
- Use packages to organize larger projects.
- Use clear, lowercase module names.
- Avoid naming files after standard modules, such as `random.py` or `json.py`.
- Prefer explicit imports.
- Avoid `from module import *`.
- Use `if __name__ == "__main__":` for executable test code.
- Use virtual environments for project dependencies.
- Keep imports near the top of the file.
- Group imports into standard library, third-party, and local imports.

Example import organization:

```python
# Standard library
import os
import json

# Third-party packages
import requests

# Local modules
from tools.calculator import add
```

A useful summary:

```text
Module  = one Python file
Package = a directory containing related modules
Import  = a way to use code from another module or package
```
