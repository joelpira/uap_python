<h2>None Value</h2>
Python has a special type, the NoneType, with a single value, None.
This is used to represent null values or nothingness.
It is not the same as False, or an empty string or 0; it is a non-value. 

```python
winner = None
print("winner:", winner)
print("winner is None:", winner is None)
print("winner is not None:", winner is not None)
print(type(winner))
print("Set winner to True")
winner = True
print("winner:", winner)
print("winner is None:", winner is None)
print("winner is not None:", winner is not None)
print(type(winner))
```

Practical use of None

```python
def parse_int(s):
    try:
        return int(s)
    except ValueError:
        return None


result = parse_int("abc")
result = parse_int(2343)
if result is None:
    print("Not a number")
else:
    print("The number is:", result)
```
