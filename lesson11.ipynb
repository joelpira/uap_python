
# Operator Overloading

Operator overloading allows user defined classes to appear to have a natural way of
using operators such as +, −, <, > or == as well as logical operators such as &
(and) and | (or).

### Compare

``` python 
q1 = Quantity(5)
q2 = Quantity(10)
q3 = q1 + q2

q1 = Quantity(5)
q2 = Quantity(10)
q3 = q1.add(q2)

```



````python
class Quantity:
    def __init__(self, value=0):
        self.value = value

    def __add__(self, other):
        new_value = self.value + other.value
        return Quantity(new_value)

    def __sub__(self, other):
        new_value = self.value - other.value
        return Quantity(new_value)

    def __str__(self):
        return 'Quantity[' + str(self.value) + ']'

q1 = Quantity(5)
q2 = Quantity(10)

print('q1 =', q1, ', q2 =', q2)
q3 = q1 + q2
print('q3 =', q3)

q1.__add__(q2)
print('q1 after add =', q1)

q1.__sub__(q2)
print('q1 after subtract =', q1)
````

<h2>Longer Explanation</h2>
Python supports **operator overloading** by defining special methods, often called **dunder methods** because they begin and end with double underscores.

For example, when Python sees:

```python
a + b
```

it roughly calls:

```python
a.__add__(b)
```

## Numerical operator overloading

| Operator | Operation | Special method | Example expression |
|---|---|---|---|
| `+` | Addition | `__add__` | `a + b` |
| `-` | Subtraction | `__sub__` | `a - b` |
| `*` | Multiplication | `__mul__` | `a * b` |
| `/` | True division | `__truediv__` | `a / b` |
| `//` | Floor division | `__floordiv__` | `a // b` |
| `%` | Modulo/remainder | `__mod__` | `a % b` |
| `**` | Exponentiation | `__pow__` | `a ** b` |
| `@` | Matrix multiplication | `__matmul__` | `a @ b` |
| `-x` | Unary negative | `__neg__` | `-x` |
| `+x` | Unary positive | `__pos__` | `+x` |
| `abs(x)` | Absolute value | `__abs__` | `abs(x)` |
| `~x` | Bitwise inversion | `__invert__` | `~x` |

## Comparison operators

| Operator | Meaning | Special method |
|---|---|---|
| `==` | Equal to | `__eq__` |
| `!=` | Not equal to | `__ne__` |
| `<` | Less than | `__lt__` |
| `<=` | Less than or equal | `__le__` |
| `>` | Greater than | `__gt__` |
| `>=` | Greater than or equal | `__ge__` |

## Example: creating a `Vector` class


````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, number):
        return Vector(self.x * number, self.y * number)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"




v1 = Vector(2, 3)
v2 = Vector(4, 5)

print(v1 + v2)  # Vector(6, 8)
print(v2 - v1)  # Vector(2, 2)
print(v1 * 3)    # Vector(6, 9)
print(v1 == v2)  # False

v1 + v2

print(v1 + v2)

print(v1.__add__(v2))

print(v1.__sub__(v2))
print(v1.__mul__(3))
print(v1.__eq__(v2))
````



## Reflected operators

Reflected methods handle cases where your object appears on the right side of an operator.

| Expression | Reflected method |
|---|---|
| `number + object` | `__radd__` |
| `number - object` | `__rsub__` |
| `number * object` | `__rmul__` |
| `number / object` | `__rtruediv__` |

Example:

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __mul__(self, number):
        return Vector(self.x * number, self.y * number)

    def __rmul__(self, number):
        return self * number

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v = Vector(2, 3)

print(v * 2)  # Vector(4, 6)
print(2 * v)  # Vector(4, 6)

````

`__rmul__` makes `2 * v` work.

Operator overloading is useful for classes representing mathematical objects, money, dates, measurements, vectors, database queries, or custom collections. The overloaded behavior should remain intuitive; for example, using `+` to combine vectors is clear, but using it to delete a file would be confusing.
