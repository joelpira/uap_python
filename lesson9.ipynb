# 🐍 OOP in Python — Introduction

Object-Oriented Programming organizes code around **objects** — bundles of data (attributes) and behavior (methods).   

## 1. Classes & Objects

| Concept | Analogy |
|---|---|
| **Class** | A blueprint / recipe |
| **Object** | A real thing built from the blueprint |
| **Attribute** | A property (name, color) |
| **Method** | An action the object can do |   

```python
class Dog:
    species = "Canine"
    name  = ""
    __hide = "This is hidden"
    _fix_value = 42
    

    def __init__(self, name, age):
    
        self.name = name
        self.age = age

    def bark(self):

        return f"{self.name} says Woof!"


rex  = Dog("Rex", 3) # __init__ method called
luna = Dog("Luna", 5)

print(rex.name)       # Rex
print(rex.bark())     # Rex says Woof!
print(Dog.species)    # Canine   
print(Dog._fix_value)
Dog._fix_value = 55
print(Dog._fix_value)
```

## 2. `self` Explained

`self` = the **current object**. Python passes it automatically:
   

```python
rex.bark()
# Python actually calls:
Dog.bark(rex)  # rex is passed as self
```

## 3. The Four Pillars of OOP

1. **Encapsulation** → hide data, expose API
2. **Inheritance** → reuse via parent/child
3. **Polymorphism** → same method, different behavior
4. **Abstraction** → simplify, hide complexity   

## 4. Encapsulation   


```python
class BankAccount:
    
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance


acc = BankAccount("Alice", 100)
acc.deposit(50)
print(acc.owner)
#print(acc.__balance)  # This will raise an AttributeError because __balance is private
print(acc.get_balance())  # 150
```

- `_name` → convention-private ("don't touch")
- `__name` → name-mangled (Python's "hiding")   


## 5. Inheritance   

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."


class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow!"


class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"

    def fetch(self, item):
        return f"{self.name} fetches {item}"


doggo = Dog("Rex")
kitty = Cat("Whiskers")
print(kitty.speak())  # Whiskers says Meow!
print(kitty.name)  # Whiskers (inherited)
```

```python
class Puppy(Dog):
    def __init__(self, name, toy):
        super().__init__(name)
        self.toy = toy


p = Puppy("Buddy", "ball")
print(p.speak())  # Buddy says Woof!
print(p.toy)  # ball
```

## 6. Polymorphism

Same method name → different behavior per object.   

```python
animals = [Cat("Whiskers"), Dog("Rex"), Animal("Blob")]

for a in animals:
    print(a.speak())
# Whiskers says Meow!
# Rex says Woof!
# ...
```

## 7. Abstraction

Hide complex internals behind a simple interface.   

```python
class Playlist:
    def __init__(self):
        self._tracks = []

    def add(self, song):
        self._tracks.append(song)

    def play(self):
        for t in self._tracks:
            print(f"▶ {t}")


p = Playlist()
p.add("Bohemian Rhapsody")
p.play()
```

## 8. Dunder (Magic) Methods   


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)


p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1)  # (1, 2)
print(p1 + p2)  # (4, 6)
print(p1 == Point(1, 2))  # True
```

## 9. Recap

| Term | Meaning |
|---|---|
| `class` | Blueprint |
| `__init__` | Constructor |
| `self` | "this object" |
| `_attr` | Convention-private |
| `__attr` | Name-mangled |
| `class Child(Parent)` | Inheritance |
| `super()` | Call parent method |
| `__str__`, `__eq__`, … | Dunder / magic methods |
