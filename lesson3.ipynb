## The If Statement and Else


``` css

if <condition-evaluating-to-boolean>:
    statement

```

``` css
if <condition-evaluating-to-boolean>:
    statement
else: 
    statement

```



Condition must be evaluate to `True` or `False`


````python
try:
    number = int(input("Enter a number: "))
    if number > 0:
        print("The number is positive.")
    elif number < 0:
        print("The number is negative.")
    else:
        print("The number is zero.")
except ValueError:
    print("Invalid input. Please enter a valid integer.")
````

### The Use of *elif*

```
elif <conditioal-evaluation-to-boolean>:
    statement

```


````python
savings = float(input("Enter your savings amount: "))
try:
    savings = float(input("Enter your savings amount: "))
    if savings < 0:
        print("Savings cannot be negative.")
    elif savings == 0:
        print("You have no savings.")
    elif savings < 1000:
        print("You have less than $1000 in savings.")   
    elif savings < 5000:
        print("You have less than $5000 in savings.")
    elif savings < 10000:
        print("You have less than $10000 in savings.")
    else:
        print("You have $10000 or more in savings.")
except ValueError:
    print("Invalid input. Please enter a numeric value.")
````

### Nesting if Statements

````python

snowing = True
# snowing = False
temp = -1
if temp < 0:
    print('It is freezing')
    if snowing:
        print('Put on boots')
        print('Time for Hot Chocolate')
    else:
        print('No snow today')
    print('Stay warm')
print('Bye')
````

### If Expressions

``` 
<result1> if <condition-is-met> else <result2>
```

````python
age = 15
status = None
# if (age > 12) and age < 20:
if 12 < age < 20:
    status = 'teenager'
else:
    status = 'not teenager'
print(status)
````

````python
status = ('teenager' if age > 12 and age < 20 else 'not teenager')
print(status)
````

## A Note on True and False

- 0, '' (empty strings), None equate to False
- Non zero, non empty strings, any object equate to True.
