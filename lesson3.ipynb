<h1>Flow of Control Using If Statement</h1>
<h2> Comparison Operators</h2>

<table><tr><th>Operator</th><th> Description</th><th> Example</th></tr>
<tr><td>==</td><td> Tests if two values are equal </td><td>3 == 3</td></tr>
<tr><td>!=</td><td> Tests that two values are not equal to each other </td><td>2 != 3</td></tr>
<tr><td>< </td><td>Tests to see if the left-hand value is less than the right-hand value</td><td> 2 < 3</td></tr>
<tr><td>> </td><td>Tests if the left-hand value is greater than the right-hand value </td><td>3 > 2</td></tr>
<tr><td><= </td><td>Tests if the left-hand value is less than or equal to the right-hand value </td><td>3 <= 4</td></tr>
<tr><td>>=</td><td> Tests if the left-hand value is greater than or equal to the right-hand value</td><td> 5 >= 4</td></tr>
</table>

<h2>Logical Operators</h2>

<table>
<tr><th>Operator </th><th>Description</th><th> Example</th></tr>
<tr><td>and </td><td>Returns True if both left and right are true </td><td> (3 < 4) and (5 > 4)</td></tr>
<tr><td>or</td><td> Returns two if either the left or the right is truce </td><td>(3 < 4) or (3 > 5)</td></tr>
<tr><td>not </td><td>Returns true if the value being tested is False</td><td> not 3 < 2</td></tr>
