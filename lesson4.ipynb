# Iteration / Looping

## While Loop

``` html
while <test-condition-is-true>:
    statement or statements

````python
count = 0
print('Starting the count.')
while count < 5:
    print(f'\t{count}', end='\n')
    count += 1
print('\nDone counting.')
````

## For Loop

``` python






for <variable-name> in range(...):
    statement
    statement

```


````python
#Loop over a set of values in range

print('\nCounting from 0 to 4 using range()')
for count in range(1,10):
    print(f'\t{count}')
print('\nDone counting.')
````

````python
# incrementing by 2
print('\nCounting from 0 to 8 by 2s using range()')
for i in range(0, 10, 2):
    # print(f'\t{i}')
    print(i, end=', ')
print('\nDone counting.')
````

````python
# anonymous loop

for _ in range(-10,5):
    print('Hello!')
````

## Break Loop Statement

````python
print('Only print code if all iterations are successful.')
num = int(input('Enter a number: '))
for i in range(1, 10):
    print(i, end=' ')
    if(num == i):
        print(f'Found {num} in the range.')
        break
print('Done searching.')

    
````

## Continue Loop Statement

````python
for i in range(0, 10):
    print(i,' ',end='')
    if i % 2 == 1:
        continue
    print('even')
print('Done counting.')
````

## For Loop with Else

optional **else**

````python
# only print code if all iterations are successful.
num = int(input('Enter a number: '))
for i in range(1, 10):
    print(i, end=' ')
    if(num == i):
        print(f'Found {num} in the range.')
        break
else:
    print(f'\n{num} was not found in the range.')
    
````

## Dice Roll Game

````python
import random
MIN = 1
MAX = 6
roll_again = 'y'
while roll_again == 'y':
    print('Rolling the dices...')
    print('The values are....')
    dice1 = random.randint(MIN, MAX)
    print(dice1)
    dice2 = random.randint(MIN, MAX)
    print(dice2)
    roll_again = input('Roll the dices again? (y / n): ')
````
