## Functions as Objects

Function that takes another function as a parameter is known as a
**<span style="color:red">higher order function</span>**.

````python
def apply(x, function):
    result = function(x)
    return result


def mult(y):
    return y * 10.0

result = apply(90, mult)
print(result)
````

### Another Example

````python
def mult_by_five(num):
    return num * 5

def square(num):
    return num * num

def add_one(num):
    return num + 1

def apply(num, func):
    return func(num)

result = 'Multiply by five: ' + str(apply(4, mult_by_five))
print(result)
result = 'Square: ' + str(apply(4, square))
print(result)
result = 'Add one: ' + str(apply(4, add_one))
print(result)
````

### Usage of Higher Order Functions

Creating a reusable piece of code that will be able to apply an
appropriate function to the data we have when that function is known.

````python
import math
def simple_tax_calculator(amount):
    return math.ceil(amount * 0.3)


## function is not known until the function is declared
def calculate_tax(salary, func):
    return func(salary)



print(str.format("Calculated tax: {}", calculate_tax(45000.0, simple_tax_calculator)))
````

## Functions Returning Functions

````python
def make_checker(s):
    if s == 'even':
        return lambda n: n%2 == 0
    elif s == 'positive':
        return lambda n: n >= 0
    elif s == 'negative':
        return lambda n: n < 0
    else:
        raise ValueError('Unknown request')


f1 = make_checker('even') #returns a function that checks if a number is even
f2 = make_checker('positive') #returns a function that checks if a number is positive
f3 = make_checker('negative') #returns a function that checks if a number is negative

print(f1(3))
print(f2(3))
print(f3(3))
````

### Another Example

````python
def make_function():
    def adder(x, y):
        return x + y
    return adder
    
f1 = make_function()
print(f1(3, 2))
print(f1(3, 3))
print(f1(3, 1))
````
