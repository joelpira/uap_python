# Python f-Strings — Lecture Notes

## What they are

An **f-string** lets you place Python expressions inside a string. Put `f` or `F` before the opening quote, and enclose each expression in braces `{}`.

```python
name = "Mira"
age = 20

message = f"{name} is {age} years old."
print(message)
# Mira is 20 years old.
```

The expression inside `{}` is evaluated, then converted to text.

## Expressions inside braces

You can use variables, calculations, function calls, and other expressions:

```python
price = 12.5
quantity = 3

print(f"Total: {price * quantity}")
# Total: 37.5

print(f"Uppercase: {"hello".upper()}")
# Uppercase: HELLO
```

For clarity, calculate complicated values before formatting:

```python
total = price * quantity
print(f"Total: {total}")
```

## Formatting values

Add a format specifier after a colon inside the braces:

```python
pi = 3.14159265

print(f"{pi:.2f}")       # 3.14 — 2 digits after the decimal
print(f"{42:05d}")       # 00042 — pad integer with zeros
print(f"{1234567:,}")    # 1,234,567 — thousands separator
print(f"{0.875:.1%}")    # 87.5% — percentage
```

Common alignment and width options:

```python
item = "apple"

print(f"{item:>10}")  # right-align in 10 characters
print(f"{item:<10}")  # left-align in 10 characters
print(f"{item:^10}")  # center in 10 characters
```

## Conversions

A conversion flag can request a particular kind of representation:

```python
text = "hello"

print(f"{text!r}")  # 'hello' — representation useful for debugging
print(f"{text!s}")  # hello — string form
print(f"{text!a}")  # 'hello' — ASCII representation
```

## Debugging with `=`

An equals sign displays the expression and its value:

```python
count = 7
print(f"{count=}")
# count=7
```

You can combine it with formatting:

```python
ratio = 0.12345
print(f"{ratio=:.2%}")
# ratio=12.35%
```

## Literal braces

Double the braces to include them as ordinary characters:

```python
name = "Mira"
print(f"{{Hello, {name}!}}")
# {Hello, Mira!}
```

## Practical example

```python
product = "Notebook"
price = 4.5
quantity = 3

print(f"{product}: {quantity} × ${price:.2f} = ${price * quantity:.2f}")
# Notebook: 3 × $4.50 = $13.50
```

## Key points

- Use `f"..."` or `f'...'` to create an f-string.
- Put expressions in `{}`.
- Use `:` to add formatting options.
- Double braces `{{` and `}}` to print literal braces.
- Don’t put untrusted input into expressions; f-strings evaluate Python code inside `{}`.
