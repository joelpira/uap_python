<h1>Python String</h1>

String
- series or sequence of characters
- Immutable - once created cannot be changed

```python
s1 = "hello"
s2 = s1.replace("h", "H")  # creates a new string
print(s1)  # hello
print(s2)  # Hello
```

<h2>Immutable</h2>

```python
s = "hello"
try:
    s[0] = "H" #strings don't support item assignment
except TypeError as e:
    print("Error: ", e)
```

<h2>String formatting</h2>
{ } as a placeholder

```python
format_string = 'Hello {}'
print(format_string.format('Phoebe'))

name = 'Adam'
age = 20

print("{} is {} years old".format(name,age))
print('{} is {} years old'.format(name,age))
```

<h3>index as a placeholde</h3>

```python
format_string = 'Hello {1} {0}, you got {2}%'
print(format_string.format('Smith','Carol','75'))
print(format_string.format('75','Carol','Smith'))
```

<h3>named values for placeholders</h3>

```python
format_string = '{artist} sang {song} in {year}'
print(format_string.format(song='Title of the Song',year='100BC',artist='RTX'))
```

<h3>indicate alignment and width within the format string</h3>

```python
print('|{:25}|'.format('25 characters width'))
```

Within this gap you can also indicate an alignment where:
• < Indicates left alignment (the default),
• > Indicates right alignment,
• ^ Indicates centered.
These follow the colon (':') and come before the size of the gap to use, for
example:

```python
print("|{:<25}|".format("left aligned"))  # The default
print("|{:>50}|".format("right aligned"))
print("|{:^25}|".format("centered"))
```

Another formatting option is to indicate that a number should be formatted with separators (such as a comma) to indicate thousands:

```python
print("{:,}".format(1234567890))
print("{:,}".format(1234567890.0))
```

<h2>String Template v3.14</h2>

```python
from string.templatelib import Template

name = "Alice"
tmpl = t"Hello, {name}!"

#access components
print(tmpl.strings)
print(tmpl.values)

#iterate to process safely 
result = ""
for item in tmpl:
    if isinstance(item, str):
        result += item
    else:
        result += item.value.upper()

print(result)
```
