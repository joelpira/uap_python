```python
import pandas as pd
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data) 


```

```python
import matplotlib.pyplot as plt

plt.plot(df["Name"], df["Age"])
plt.title("Age of Individuals")
plt.xlabel("Name")
plt.ylabel("Age")
plt.show()
```
